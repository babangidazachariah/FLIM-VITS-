```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import make_interp_spline

# Data
flow_rates = [100, 200, 300, 400, 500, 600, 700, 800, 900, 1000]
TPPD = [52.26, 58.27, 108.25, 106, 101.69, 96.95, 98.12, 110, 143.42, 158.25]
I2IPPD = [51.93, 53.56, 67.27, 88.42, 71.56, 61, 64.19, 67.81, 60.69, 68.54]

# Generate smooth curves
x_new = np.linspace(min(flow_rates), max(flow_rates), 300)

spl_TPPD = make_interp_spline(flow_rates, TPPD, k=3)  # type: BSpline
smooth_TPPD = spl_TPPD(x_new)

spl_I2IPPD = make_interp_spline(flow_rates, I2IPPD, k=3)  # type: BSpline
smooth_I2IPPD = spl_I2IPPD(x_new)

# Plotting
plt.figure(figsize=(15, 10))  # Set figure size for better resolution
plt.plot(x_new, smooth_TPPD, label='TPC')
plt.plot(x_new, smooth_I2IPPD, label='I2IPC')
plt.scatter(flow_rates, TPPD, marker='o', color='blue')
plt.scatter(flow_rates, I2IPPD, marker='s', color='orange')
plt.xlabel('Flow Rates (vph)')
plt.ylabel('Delay (s)')
plt.title('Priority Delay Plot')
plt.legend()
plt.grid(True)
plt.show()
```
